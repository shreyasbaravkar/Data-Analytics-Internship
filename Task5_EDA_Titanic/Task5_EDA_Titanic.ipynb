```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 

# Display plots inside the notebook 
%matplotlib inline 

# Set a clean style for plots 
sns.set_style("whitegrid")
```

```python
# Load the dataser
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
# Display the first 5 rows
df.head()
```

```python
# Basic information about the dataset
df.info()
```

```python
# Statistical summary of numerical columns 
df.describe()
```

```python
# Check for missing values 
df.isnull().sum()
```

```python
# Count unique values in categorical columns 
print(df["Sex"].value_counts())
print()
print(df["Embarked"].value_counts())
print()
print(df["Survived"].value_counts())
print()
print(df["Pclass"].value_counts())
```

```python
# Histograms for all numerical columns 
df.hist(figsize=(12,10),bins=20)
plt.tight_layout()
plt.show()
```

```python
# Boxplots for Age and Fare
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.boxplot(y=df["Age"])
plt.title("Boxplot of Age")

plt.subplot(1,2,2)
sns.boxplot(y=df["Fare"])
plt.title("Boxplot of Fare")

plt.tight_layout()
plt.show()

```

```python
# Correlation heatmap
plt.figure(figsize=(10,8))

numeric_df = df.select_dtypes(include=["number"])

sns.heatmap(numeric_df.corr(),annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Pairplot of selected numerical features
sns.pairplot(
    df[["Survived","Age","Fare", "Pclass"]],
    hue="Survived"
)
```

```python
# Survival Count
plt.figure(figsize=(6,4))
sns.countplot(x="Survived", data=df)
plt.title("Survival Count")
plt.show()
```

```python
# Survival by gender 
plt.figure(figsize=(6,4))
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by gender")
plt.show()
```

```python
# Survival by Passenger Class
plt.figure(figsize=(6,4))
sns.countplot(x="Pclass",hue="Survived",data=df)
plt.title("Surivival by Passenger Class")
plt.show()
```

```python
# Key Insights from Exploratory Data Analysis

#1. The dataset contains missing values mainly in the Age, Cabin, and Embarked columns.

#2. The Fare column is highly right-skewed and contains several outliers.

#3. Most passengers were in 3rd class, while fewer traveled in 1st and 2nd class.

#4. Female passengers had a much higher survival rate than male passengers.

#5. Passengers in 1st class had a better chance of survival compared to those in 2nd and 3rd class.

#6. The correlation heatmap showed that Fare has a positive correlation with Survived, while Pclass has a negative correlation with Survived.

#7. Pairplots helped visualize the relationships between Age, Fare, Pclass, and Survived, showing distinct patterns based on survival.

#8. Countplots clearly illustrated the distribution of passengers and survival across different categories.
```
